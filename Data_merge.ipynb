```python
import csv

def tsv_to_csv(tsv_file, csv_file):
    with open(tsv_file, 'r', newline='') as tsv_in, open(csv_file, 'w', newline='') as csv_out:
        tsv_reader = csv.reader(tsv_in, delimiter='\t')
        csv_writer = csv.writer(csv_out)
        
        for row in tsv_reader:
            csv_writer.writerow(row)

# Example usage
tsv_to_csv('TCGA_LUAD_LUAS_Nor.tsv', 'TCGA_LUAD_LUAS_Nor.csv')
```

```python
import pandas as pd

# Load the first dataset
file1 = 'Prepared_Data_1_TCGA__A-S-N_Final.csv'
df1 = pd.read_csv(file1)

# Load the second dataset
file2 = 'Prepared_Data_2_ICGC__A-N_Final.csv'
df2 = pd.read_csv(file2)

# Identify common genes
common_genes = set(df1.columns).intersection(set(df2.columns))
common_genes.remove('Sample_ID')

# Filter both datasets to include only the common genes
df1_filtered = df1[['Sample_ID'] + list(common_genes)]
df2_filtered = df2[['Sample_ID'] + list(common_genes)]

# Merge the datasets on Sample_ID
merged_df = pd.merge(df1_filtered, df2_filtered, on='Sample_ID', how='outer', suffixes=('_file1', '_file2'))

# Save the merged dataset to a new CSV file
merged_df.to_csv('Prepared_Data_TCGA_ICGC_Commons.csv', index=False)

print("Merged dataset saved to 'Prepared_Data_TCGA_ICGC_Commons.csv'.")
```

# Merging Datasets based on Common Genes


```python
import pandas as pd

# Load the two CSV files
df1 = pd.read_csv('Prepared_Data_1_TCGA__A-S-N_Final.csv')
df2 = pd.read_csv('Prepared_Data_2_ICGC__A-N_Final.csv')

# Identify common genes (columns)
common_genes = df1.columns.intersection(df2.columns)
common_genes = common_genes.drop('Sample_ID')  # Drop 'sampleid' from the common columns

# Select the common columns in both dataframes
df1_common = df1[['Sample_ID'] + list(common_genes)]
df2_common = df2[['Sample_ID'] + list(common_genes)]

# Concatenate the dataframes
merged_df = pd.concat([df1_common, df2_common])

# Save the merged dataframe to a new CSV file
merged_df.to_csv('TCGA_ICGC_Commond_Genes_Merged.csv', index=False)
```

# Datasets Merged based on All Genes

```python
import pandas as pd

# Reading CSV files
df1 = pd.read_csv('E:Prepared_Data_1_TCGA__A-S-N_Final.csv')
df2 = pd.read_csv('E:Prepared_Data_2_ICGC__A-N_Final.csv')

# Merging DataFrames
merged_df = pd.merge(df1, df2, on='Sample_ID', how='outer')

# Get all unique columns from both dataframes except 'Sample_ID'
all_columns = list(set(df1.columns[1:]).union(set(df2.columns[1:])))

# Add 'Sample_ID' as the first column and sort the columns
sorted_columns = ['Sample_ID'] + sorted(all_columns)

# Create an empty dataframe with all the required columns
empty_df = pd.DataFrame(columns=sorted_columns)

# Populate the empty dataframe with data from the merged dataframe
merged_df = pd.concat([empty_df, merged_df], ignore_index=True)

# Filling missing values with NaN
merged_df = merged_df.fillna('NaN')

# Saving the merged DataFrame to a CSV file
merged_df.to_csv('E:merged_data.csv', index=False)

print(merged_df)
```
